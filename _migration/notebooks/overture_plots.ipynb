```python
county_url = "https://www2.census.gov/geo/tiger/TIGER2023/COUNTY/tl_2023_us_county.zip"
```

```python
county = gpd.read_file(county_url)
```

```python
county.columns
```

```python
exclude_fips = ['02','15','60',
 '66',
 '69',
 '72',
 '78']
```

```python
cont_us_counties = county.query("STATEFP not in @exclude_fips")
```

```python
cont_us_counties.GEOID.unique().shape
```

```python
cont_us_counties.shape
```

```python
cont_us_counties.plot()
```

```python
joined_places = gpd.overlay(result,cont_us_counties.to_crs('EPSG:4326'))
```

```python
point_cnt = joined_places.groupby('GEOID').agg({'main_cat':'count'}).reset_index().rename(columns={'main_cat':'cnt'})
```

```python
cont_us_counties_ovm = cont_us_counties.merge(point_cnt,on='GEOID')
```

```python
cont_us_counties_ovm.__geo_interface__
```

```python
cont_us_counties_ovm.sort_values('')
```

```python
alt_data  = alt.InlineData(values = cont_us_counties_ovm.to_json(), #geopandas to geojson
                       # root object type is "FeatureCollection" but we need its features
                       format = alt.DataFormat(property='features',type='json'))
```

```python
cont_us_counties_ovm = gpd.read_file("../data/overture/cont_us_counties_ovm.geojson")
```

```python
source = cont_us_counties_ovm[['GEOID','cnt']].rename(columns={'GEOID':'id'})
```

```python
counties = alt.topo_feature("https://cdn.jsdelivr.net/npm/us-atlas@3/counties-10m.json",feature="counties")

choropleth_chart = alt.Chart(counties).mark_geoshape().encode(
    color=alt.Color('cnt:Q',
                    scale=alt.Scale(type='log',#bins=[0, 500, 1000, 1500, 2000000],
                                    scheme=alt.SchemeParams(name='greenblue')))
).transform_lookup(
    lookup='id',
    from_=alt.LookupData(source, 'id', ['cnt'])
).project(
    type='albersUsa'
).properties(
    width=1000,
    height=600
)
```
